```python
import pyspark
from pyspark import SparkContext

# Si ya existe un contexto anterior, lo cerramos para evitar errores
if 'sc' in globals():
    sc.stop()

sc = SparkContext.getOrCreate()
data = sc.textFile("data/calidad_aire_datos_meteo_mes.csv")

# Extraer y eliminar la cabecera
header = data.first()
print("Cabecera:", header)  # Verifica que sea correcta
# Filtrar la cabecera
data_no_header = data.filter(lambda row: row != header)

print("Datos cargados correctamente. Número total de filas:", data_no_header.count())
```
