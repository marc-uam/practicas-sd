```python
import pyspark

# Inicializo y cargo los datos
if 'sc' in globals():
    sc.stop()

sc = pyspark.SparkContext('local[*]')

data = sc.textFile('data/calidad_aire_datos_meteo_mes.csv')  

header = data.first()
data_no_header = data.filter(lambda row: row != header)

print("Datos cargados correctamente")
print("Número total de filas:", data_no_header.count())

# ejercicio primero
magnitud_idx = 3
valid_indices = list(range(9, 56, 2))   # v01, v02, ..., v24

parsed = data_no_header.map(lambda line: line.split(';'))

temp_records = parsed.filter(lambda row: row[magnitud_idx] == '83')

valid_temp = temp_records.filter(
    lambda row: any(row[i] == 'V' for i in valid_indices)
)

count = valid_temp.count()
print(f"El numero de registros de temperatura con al menos un dato válido son: {count}")
```
