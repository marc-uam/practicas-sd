```python
import pyspark

# Inicio la carga de datps
if 'sc' in globals():
    sc.stop()

sc = pyspark.SparkContext('local[*]')

data = sc.textFile('data/calidad_aire_datos_meteo_mes.csv')        

header = data.first()
data_no_header = data.filter(lambda row: row != header)

print("Datos cargados correctamente")
print("Número total de filas:", data_no_header.count())

# ejercicio2
def get_date(row):
    year = row[5]
    month = row[6].zfill(2)
    day = row[7].zfill(2)
    return f"{year}-{month}-{day}"

# Ffiltra solo la temperatura
temp_records = data_no_header.map(lambda line: line.split(';')) \
                            .filter(lambda row: row[3] == '83')

# extrae las temperaturas validas
def extract_valid_temps(row):
    date = get_date(row)
    valid_temps = []
    for i in range(24):
        h_idx = 8 + 2 * i      # h01 = 8, h02 = 10, ...
        v_idx = 9 + 2 * i      # v01 = 9, v02 = 11, ...
        
        valor = row[h_idx].strip()
        if row[v_idx] == 'V' and valor != '' and valor != '..':
            try:
                valor_float = float(valor.replace(',', '.'))
                valid_temps.append((date, valor_float))
            except:
                pass  # ignorar valores que no se puedan convertir
    return valid_temps

date_temp_pairs = temp_records.flatMap(extract_valid_temps)

# Temperatura máxima por día
max_per_day = date_temp_pairs.reduceByKey(lambda a, b: max(a, b))

# Ordenamos y mostramos
sorted_results = max_per_day.sortByKey()

print("Temperatura maxima del dia (magnitud 83)")
results = sorted_results.collect()
for date, max_temp in results:
    print(f"{date}: {max_temp:.2f} °C")

print(f"\nTotal de días procesados: {len(results)}")
```
